```python
from bunq.sdk.model.generated.endpoint import MonetaryAccountBankApiObject
from bunq.sdk.model.generated.endpoint import PaymentApiObject
from bunq import Pagination


# Create pagination
pagination = Pagination()
pagination.count = 10

# get a list of all accounts
monetary_accounts = MonetaryAccountBankApiObject.list(params=pagination.url_params_count_only).value

for account in monetary_accounts:
    print(f"Account ID: {account.id_}")
    print(f"Description: {account.description}")
    print(f"Balance: {account.balance.value} {account.balance.currency}")
    print(f"IBAN: {account.alias[0].value if account.alias else 'No IBAN'}")
    print("---")
```
